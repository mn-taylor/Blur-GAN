```python
from flow_utils import *
import matplotlib.pyplot as plt
```

```python
num_epochs = 30
x = np.arange(0, num_epochs)
t, e = cosine_flow_funcs(num_epochs-1, cool_down=num_epochs*2/5)

y = np.array([t(i) for i in x])
y_e = np.array([1 * e(i) for i in x])

plt.plot(x, y)
plt.fill_between(x, y-y_e, y + y_e, color="red", alpha=0.2)
print(t(20))
```
